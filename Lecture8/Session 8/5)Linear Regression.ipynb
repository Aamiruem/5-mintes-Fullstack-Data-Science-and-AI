```python
from sklearn.datasets import make_regression

x, y = make_regression(n_features=1, noise=5, n_samples=10000, random_state=1)
import matplotlib.pyplot as plt  # noqa: E402

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

🧠 1. from sklearn.model_selection import train_test_split
sklearn

Machine Learning library.

model_selection

Module that contains tools for:

splitting data

cross-validation

model evaluation

import

Bring something into your program.

train_test_split

A function that divides your dataset into:

training data → used to learn

testing data → used to evaluate

✂ 2. X_train, X_test, y_train, y_test = ...

These four variables receive the output:

Variable	Meaning
X_train	Input data for training
X_test	Input data for testing
y_train	Output for training
y_test	Output for testing
🧮 3. train_test_split(x, y, test_size=0.2, random_state=0)
x

Your input features.

y

Your output labels.

test_size=0.2

Means:

20% data → testing
80% data → training

Since you had 10,000 samples:

Set	Size
Training	8000
Testing	2000
random_state=0

Fixes randomness so the split is same every run.

🧩 What this line really does

Before:

x → 10000 samples
y → 10000 samples


After:

X_train → 8000 samples
X_test  → 2000 samples
y_train → 8000 samples
y_test  → 2000 samples

🧠 Why we do this?

Because a model must be:

trained on known data

tested on unseen data

This tells us if the model actually learned or just memorized.

```python
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
X_test
```

```python
regressor.predict(X_test)
```

```python
regressor.score(X_test, y_test)
```

```python
regressor.coef_
```

```python
regressor.intercept_
```

```python
y_pred = regressor.predict(x)
y_pred
```

```python
plt.scatter(x, y, label="Actual")
plt.scatter(x, y_pred, label="Predicted")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```
