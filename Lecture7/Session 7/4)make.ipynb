```python
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_regression
```

```python
x, y = make_regression(
    n_features=1, noise=5, n_samples=10000, random_state=0
)  # means we have 1 independent variable noise = 5 # means we are adding some random noise to the data points n_samples = 10000 means we are generating 10000 data points
```

```python
x
```

```python
y
```

```python
import matplotlib.pyplot as plt

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Scatter plot of generated data")
plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.colorbar()
plt.show()
```

```python
from sklearn.datasets import make_classification
```

```python
x, y = make_classification(n_classes=2, n_features=5, n_samples=10000)
plt.scatter(
    x[:, 0], x[:, 1], c=y
)  # n_features=5 means we have 5 independent variables n_classes=2 means we have 2 classes (0 and 1) n_samples=10000 means we are generating 10000 data points (x[:,0] means we are taking all the rows and first column) (x[:,1] means we are taking all the rows and second column) (c=y means we are coloring the points based on their class)
```
